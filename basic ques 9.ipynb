```python
import pandas as pd

df = pd.DataFrame({
    'City': ['New York', 'London', 'New York', 'London'],
    'Sales': [1000, 2000, 1500, 3000]
})

city_totals = df.groupby('City')['Sales'].sum()

# This line will throw a KeyError
try:
    print(city_totals['City'])
except KeyError as e:
    print(f"Error: {e}")

print("\nThe actual result is:")
print(city_totals)
```
