```python
import pandas as pd

df = pd.DataFrame({
    'Product': ['Laptop', 'Shirt', 'Phone', 'Apple'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Fruit']
})
# The ~ symbol is the "bitwise NOT" for Pandas
non_electronics = df[~(df['Category'] == 'Electronics')]
```

```python
print(non_electronics)
```
