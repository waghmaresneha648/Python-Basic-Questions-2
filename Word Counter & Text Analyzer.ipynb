```python
text = input("Enter a sentence or paragraph:\n")

# Split text by spaces to count words
words = text.split()
word_count = len(words)
char_count = len(text)
# Count sentences by looking for punctuation marks
sentence_count = text.count(".") + text.count("!") + text.count("?")

print("\n=== Text Stats ===")
print(f"Total Words: {word_count}")
print(f"Total Characters: {char_count}")
print(f"Estimated Sentences: {sentence_count}")
```
