```python
def count_vowels(text):
    # Define what vowels are (both lowercase and uppercase)
    vowels = "aeiouAEIOU"

    # Keep track of counts
    total_vowels = 0
    vowel_counts = {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}

    # Loop through every character in the text
    for char in text:
        if char in vowels:
            total_vowels += 1
            # Convert to lowercase to group 'A' and 'a' together
            vowel_counts[char.lower()] += 1

    return total_vowels, vowel_counts


# Get sentence/paragraph input from the user
user_input = input("Enter a sentence or paragraph:\n")

# Run the function
total, breakdown = count_vowels(user_input)

# Display the results
print("\n=== Vowel Analysis ===")
print(f"Total vowels found: {total}")
print("Breakdown by vowel:")
for vowel, count in breakdown.items():
    print(f"  {vowel.upper()}: {count}")
```
