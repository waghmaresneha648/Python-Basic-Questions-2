```python
#Stemming  (Chopping off endings)
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running", "jumped", "cats", "studies"]

for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```

```python
#POS Tagging 
import nltk


# 1. Download Tokenizer data (Required for breaking sentences into words)
nltk.download('punkt')
nltk.download('punkt_tab')

# 2. Download POS Tagger models (Required for POS Tagging)
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# 3. Download WordNet database (Required for Lemmatization)
nltk.download('wordnet')
from nltk.tokenize import word_tokenize

text = "The quick brown fox jumps over the lazy dog"

# 1. Break sentence into words
tokens = word_tokenize(text)

# 2. Tag each word
tags = nltk.pos_tag(tokens)

print(tags)
```

```python
#Lemmatization Code
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Tell it if the word is a Noun ('n') or Verb ('v')
print("rocks  ->", lemmatizer.lemmatize("rocks", pos="n"))
print("better ->", lemmatizer.lemmatize("better", pos="a"))  # 'a' for adjective
print("running->", lemmatizer.lemmatize("running", pos="v"))
print("ate    ->", lemmatizer.lemmatize("ate", pos="v"))
```
