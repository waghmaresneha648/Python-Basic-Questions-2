```python
def is_perfect_rotation(s1, s2):
    # If lengths are different, they can't be rotations
    if len(s1) != len(s2):
        return False
    
    # The "Magic" logic:
    # 'python' + 'python' = 'pythonpython'
    # 'onpyth' is inside 'pythONPYTHon'
    return s2 in (s1 + s1)

# Test cases
print(is_perfect_rotation("python", "onpyth")) # Output: True
print(is_perfect_rotation("apple", "leapp"))   # Output: True
print(is_perfect_rotation("code", "doce"))     # Output: False
```
