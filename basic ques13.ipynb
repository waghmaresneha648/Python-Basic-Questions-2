```python
import pandas as pd

df = pd.DataFrame({
    'Day': [1, 2, 3, 4, 5],
    'Status': ['Profit', 'Profit', 'Loss', 'Loss', 'Profit']
})

# Using .shift() to compare current row to previous row
df['Change'] = df['Status'] != df['Status'].shift()

print(df)
```

```python
df['Change'] = df['Status'] != df['Status'].shift().fillna(df['Status'].iloc[0])
print(df)
# Option B: Just manually set the first row
#df.loc[0, 'Change'] = False
```
