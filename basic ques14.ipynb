```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Dept': ['IT', 'IT', 'HR', 'HR', 'IT'],
    'Salary': [120000, 110000, 80000, 90000, 40000]
})

# 1. Calculate the mean salary for the department of EACH row
df['Dept_Avg'] = df.groupby('Dept')['Salary'].transform('mean')

# 2. Filter the original DataFrame
result = df[df['Dept_Avg'] > 100000]

print(result)
```
