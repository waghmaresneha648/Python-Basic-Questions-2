```python
#a function to track data cleaning "flags." If a user doesn't provide a list of existing flags, the function should start with an empty list and add a new flag to it.
```

```python
def add_flag(flag_name, flag_list=[]):
    """
    A function that adds a 'flag' to a list. 
    Uses a default empty list if none is provided.
    """
    flag_list.append(flag_name)
    return flag_list

# First call: We expect ['Missing_Value']
print("Call 1:", add_flag("Missing_Value"))

# Second call: We expect ['Outlier']... or do we?
print("Call 2:", add_flag("Outlier"))

# Third call: Let's see what happens next
print("Call 3:", add_flag("Duplicate_Row"))
```
