```python
phrase = input("Enter a word or phrase: ")

# Remove spaces and convert to lowercase for accurate checking
cleaned_phrase = phrase.replace(" ", "").lower()

# Reverse the string using slice [::-1]
reversed_phrase = cleaned_phrase[::-1]

if cleaned_phrase == reversed_phrase:
    print(f'"{phrase}" IS a palindrome!')
else:
    print(f'"{phrase}" is NOT a palindrome.')
```
