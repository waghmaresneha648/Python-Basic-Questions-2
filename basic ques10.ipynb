```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Deals': [10, np.nan, 15, 5]
})

# Sorting descending
result = df.sort_values('Deals', ascending=False)

print(result)
```

```python
# Force NaNs to the top
result = df.sort_values('Deals', ascending=False, na_position='first')
```

```python
print(result)
```
