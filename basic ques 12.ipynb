```python
import pandas as pd

df = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'City': ['Paris', 'Berlin', 'Tokyo', 'London']
})

priority_cities = ['Paris', 'Tokyo']

# The WRONG way:
# result = df[df['City'] in priority_cities] --> Raises ValueError

# The CORRECT way:
result = df[df['City'].isin(priority_cities)]

print("Filtered DataFrame:")
print(result)
```
