```python
#ou have a DataFrame showing sales for different products across two years. The data is structured with a multi-level row index (Year and Product).
```

```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    (2024, 'Apples'), (2024, 'Bananas'),
    (2025, 'Apples'), (2025, 'Bananas')
], names=['Year', 'Product'])

df = pd.DataFrame({'Sales': [100, 150, 120, 180]}, index=index)

# Unstacking the DataFrame
result = df.unstack()

print(result)
```
