```python
#The Case of the Duplicate ID Scenario:You are given a list of sales transactions. Each transaction has a User_ID and a Spend amount. However, some IDs are strings ('101') and some are integers (101).
```

```python
import pandas as pd
data= { 
    'User_id':[101,'101',102,101,104],
    'spend':[80,50,20,70,80]
}
df =pd.DataFrame(data)
print(df)
total_spend=df.groupby('User_id')['spend'].sum()
print(total_spend)
```
