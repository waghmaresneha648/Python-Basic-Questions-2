```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Anna', 'Ben', 'Chris'],
    'City': ['  new york  ', 'LONDON', 'paris  ']
})

# .strip() removes the extra spaces
# .title() capitalizes the first letter of each word
df['City'] = df['City'].str.strip().str.title()

print(df)
```
