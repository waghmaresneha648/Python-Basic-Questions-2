```python
import pandas as pd

prices = pd.Series([100, 200, 300], index=['Apple', 'Banana', 'Cherry'])
discounts = pd.Series([5, 10, 15], index=['Apple', 'Cherry', 'Dates'])
final_cost = prices - discounts
print(final_cost)
# This treats missing entries as 0 instead of NaN
final_cost = prices.sub(discounts, fill_value=0)
print("--------")
print(final_cost)
```
